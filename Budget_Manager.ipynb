```python

import pandas as pd

def budget_manager():    

        
        name = input("What's your name?") #input name 
        balance = float(input("Enter your balance: ")) #input balance
       
        print("Hello", name + "!") #print name 
        print("Your initial balance is:", balance) #print initial balance

        expense_types = []
        expense_amounts = []

        for i in range(5):
           
                 
            Type = str(input(f"Enter the type of expense# {i+1}:")) #input all the 3 expense type
            Expense = float(input(f"Enter your expense# {i+1}:")) #input all the 3 expense

            expense_types.append(Type)
            expense_amounts.append(Expense)

            print(f"Your expense type# {i+1}", Type)
            print(f"Your expense amount# {i+1}", Expense) 

    
            if Expense > balance:
             print(f"Insufficient balance# {i+1}")
             return(i+1)

            else:
             print(f"Balance sufficient# {i+1}") 

        df = pd.DataFrame({
     "Expense Type": expense_types,
     "Expense Amount": expense_amounts
     })

        
        from IPython.display import display
        total_spent = sum(expense_amounts)
        remaining_balance = balance - total_spent

        display(
    df.style
    .format({"Expense Amount": "₹{:,.2f}"})
    .set_caption("Expense Summary")
    .set_table_styles([
        {"selector": "thead th", "props": [
            ("background-color", "#1f4e78"),
            ("color", "white"),
            ("text-align", "left"),
            ("padding", "10px"),
        ]},
        {"selector": "tbody td", "props": [
            ("padding", "8px 10px"),
            ("border-bottom", "1px solid #d9e2ec"),
        ]},
        {"selector": "tbody tr:nth-child(even)", "props": [
            ("background-color", "#d8121cd3"),
        ]},
    ])
)

        summary = pd.DataFrame({
    "Summary": ["Starting Balance", "Total Spent", "Remaining Balance"],
    "Amount": [balance, total_spent, remaining_balance],
})

        display(
    summary.style
    .format({"Amount": "₹{:,.2f}"})
    .hide(axis="index")
    .set_table_styles([
        {"selector": "thead th", "props": [
            ("background-color", "#22781f"),
            ("color", "white"),
            ("text-align", "left"),
            ("padding", "10px"),
        ]},
        {"selector": "tbody td", "props": [
            ("padding", "8px 10px"),
            ("border-bottom", "1px solid #d9e2ec"),
        ]},
    ])
)
            
            
           
budget_manager()
  

    






 
     
                             

                
                

```
