```python

import pandas as pd
from IPython.display import display

def budget_manager():

    name = input("What's your name? ")
    balance = float(input("Enter your starting balance: "))
    starting_balance = balance

    print("Hello", name + "!")
    print("Your initial balance is:", balance)

    expense_types = []
    expense_amounts = []
    expense_status = []

    for i in range(5):

        expense_type = input(f"Enter the type of expense #{i+1}: ")
        expense = float(input(f"Enter your expense #{i+1}: "))

        expense_types.append(expense_type)
        expense_amounts.append(expense)

        print(f"Your expense type #{i+1}: {expense_type}")
        print(f"Your expense amount #{i+1}: ₹{expense:.2f}")

        if expense > balance:
            print(f"Insufficient balance #{i+1}")
            expense_status.append("Insufficient")
        else:
            print(f"Balance sufficient #{i+1}")
            balance -= expense
            expense_status.append("Sufficient")

    # Create the expense DataFrame
    df = pd.DataFrame({
        "Expense Type": expense_types,
        "Expense Amount": expense_amounts,
        "Status": expense_status
    })

    # Display the expense table
    display(
        df.style
        .format({"Expense Amount": "₹{:,.2f}"})
        .apply(
            lambda row: [
                "color: red"
                if row["Status"] == "Insufficient"
                else ""
                for _ in row
            ],
            axis=1
        )
    )

    # Calculate total successful expenses
    total_spent = sum(
        amount
        for amount, status in zip(expense_amounts, expense_status)
        if status == "Sufficient"
    )

    # Create the summary DataFrame
    summary_df = pd.DataFrame({
        "Item": [
            "Starting Balance",
            "Total Spent",
            "Remaining Balance"
        ],
        "Amount": [
            starting_balance,
            total_spent,
            balance
        ]
    })

    # Display the summary table
    display(
        summary_df.style.format({
            "Amount": "₹{:,.2f}"
        })
    )


budget_manager()
```
